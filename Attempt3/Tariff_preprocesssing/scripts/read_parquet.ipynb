# Read Parquet File

This notebook reads and displays the parquet file from the tariffs dataset.

```python
import pandas as pd
import os
```

```python
# Set the path to the parquet file
parquet_path = '../../parquet_tariffs/provider_id=23/year=2025/part-0.parquet'

# Check if file exists
if os.path.exists(parquet_path):
    print(f"File found: {parquet_path}")
else:
    print(f"File not found: {parquet_path}")
    print(f"Current directory: {os.getcwd()}")
```

```python
# Read the parquet file
df = pd.read_parquet(parquet_path)

# Display basic information
print(f"Shape: {df.shape}")
print(f"\nDataFrame Info:")
print(df.info())
```

```python
# Display rows where month is 4
print(f"\nRows where month is 4:")
df[df['month'] == 4]

#Find NaN values in column elec_piek_per_kwh
nan_count = df['elec_piek_per_kwh'].isna().sum()
print(f"\nNumber of NaN values in 'elec_piek_per_kwh': {nan_count}")
```

```python
# Display summary statistics
df.describe()
```

```python
# Display column names and data types
print("Column Names and Data Types:")
print(df.dtypes)
```

## ENTSOE Prices Data

Read and display the ENTSOE prices parquet file.

```python
# Set the path to the ENTSOE prices parquet file
entsoe_path = '../../../Visualizations/entsoe_prices_NL_2025_2026_combined.parquet'

# Check if file exists
if os.path.exists(entsoe_path):
    print(f"File found: {entsoe_path}")
else:
    print(f"File not found: {entsoe_path}")
    print(f"Current directory: {os.getcwd()}")
```

```python
# Read the ENTSOE prices parquet file
df_entsoe = pd.read_parquet(entsoe_path)

# Display basic information
print(f"Shape: {df_entsoe.shape}")
print(f"\nDataFrame Info:")
print(df_entsoe.info())
```

```python
# Display the first few rows
print("First 10 rows:")
df_entsoe.head(10)
```

```python
# Display summary statistics
print("Summary Statistics:")
df_entsoe.describe()
```

```python
# Display column names and data types
print("Column Names and Data Types:")
print(df_entsoe.dtypes)
```

## Filter and Save 2025 Data

```python
# Filter data for 2025 using the ts_utc column
df_2025 = df_entsoe[df_entsoe['ts_utc'].dt.year == 2025]

print(f"Original shape: {df_entsoe.shape}")
print(f"2025 filtered shape: {df_2025.shape}")
print(f"\nFirst few rows of 2025 data:")
df_2025.head()
```

```python
# Write filtered 2025 data to new parquet file
output_path = '../../../Visualizations/entsoe_prices_NL_2025.parquet'

# Drop the datetime column if it exists (keep only ts_utc and price)
df_2025_clean = df_2025.drop(columns=['datetime'], errors='ignore')

df_2025_clean.to_parquet(output_path, engine='pyarrow')

print(f"Successfully saved 2025 data to: {output_path}")
print(f"Saved shape: {df_2025_clean.shape}")
print(f"Columns: {list(df_2025_clean.columns)}")
print(f"File size: {os.path.getsize(output_path) / (1024*1024):.2f} MB")
```
